La codificación por conteo reemplaza cada categoría por el número de observaciones que la contienen, la codificación por frecuencia usa la proporción correspondiente. Ambas mantienen una sola columna, aunque categorías distintas con igual conteo pueden quedar representadas por el mismo valor.

Usaremos un conjunto de datos meteorológicos para construir categorías de pronóstico y observar cómo sus conteos y proporciones se convierten en variables numéricas. Al final, compararemos esta idea con *Bag of Words*, que representa cada texto mediante el conteo de sus palabras.

```python
import pandas as pd

data = pd.read_csv('temps.txt', sep=r'\s+', header=None)  # Registros meteorológicos
data.columns = ['Temperatura máxima', 'Temperatura mínima', 'HDD', 'CDD', 'Rain']
d = ['Sábado', 'Domingo', 'Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes']
dias = [d[i % 7] for i in range(len(data))]
catego = ['Soleado', 'Mayormente soleado', 'Parcialmente nublado', 'Mayormente nublado', 'Nublado', 'Lluvioso']
prono = [catego[-1] if data['Rain'][i] > 0 else catego[int(round(1.2 * (data['HDD'][i] - data['CDD'][i]) + 2))] for i in range(len(data))]
df = pd.DataFrame(columns=['Días', 'Pronóstico', 'Temperatura máxima', 'Temperatura mínima'], index=range(len(data)))
df.iloc[:, 2:4] = data.iloc[:, 0:2]
df['Días'] = dias
df['Pronóstico'] = prono

txt1 = 'Pedro Perez Pereira, pobre pintor portuguez, pinta preciosos paisajes para poder partir para Paris.'
txt2 = 'Preciosos gatos y perros con mancha quiero. Partir de izquierda a derecha para poder llegar a algun lugar.'
txt3 = 'En un lugar de la Mancha, cuyo nombre no quiero acordarme...'
```

#### Otros casos de uso del conteo de frecuencias en variables categóricas

- **Detección de valores atípicos o categorías raras:** Las categorías con baja frecuencia pueden ser agrupadas como "otros" o tratadas de forma especial.
- **Modelos de scoring crediticio:** En variables como ocupación, ciudad o producto, el conteo de frecuencia ayuda a reducir la dimensionalidad y a identificar categorías relevantes.
- **Sistemas de recomendación:** El conteo de frecuencia de productos, géneros o autores puede ser útil como variable de entrada para modelos de recomendación.
- **Análisis de churn o abandono:** La frecuencia de ciertas acciones o categorías (por ejemplo, tipo de queja, canal de contacto) puede ser predictiva del abandono de clientes.
- **Preprocesamiento para modelos de árboles:** Los algoritmos como Random Forest o XGBoost pueden beneficiarse del conteo de frecuencia en lugar de One-Hot cuando hay muchas categorías.
- **Agrupación de categorías poco frecuentes:** Permite identificar y agrupar categorías poco representadas para evitar ruido en el modelo.

## Ejercicio aplicado: conteo de categorías con ventas de Melbourne

Aplicaremos **count encoding** (codificación por conteo) al conjunto real de ventas de propiedades de Melbourne incluido en `Data/melb_data.csv`. La variable `Suburb` es nominal, por lo que sus categorías no tienen un orden natural. Reemplazaremos cada suburbio por el número de propiedades de ese suburbio que aparecen en entrenamiento.

La codificación resume la frecuencia de cada categoría en una sola característica numérica, pero no representa directamente el precio ni hace que categorías distintas sean identificables si tienen el mismo conteo.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df_melbourne = pd.read_csv('Data/melb_data.csv')

print(f'Filas: {len(df_melbourne):,}')
print(f'Suburbios distintos: {df_melbourne["Suburb"].nunique()}')
print(f'Valores faltantes en Suburb: {df_melbourne["Suburb"].isna().sum()}')
df_melbourne[['Suburb', 'Price']].head()
```

```python
train_melbourne, test_melbourne = train_test_split(
    df_melbourne, test_size=0.25, random_state=42
)

# Ajustar los conteos únicamente con los datos de entrenamiento.
conteos_suburbio = train_melbourne['Suburb'].value_counts()

train_melbourne = train_melbourne.copy()
test_melbourne = test_melbourne.copy()
```

#### Ventajas y desventajas observadas del conteo

**Ventajas de la codificación por conteo**

- Produce una sola columna por variable, incluso si hay cientos o miles de categorías, en este conjunto hay 314 suburbios y one-hot genera una columna por categoría vista en entrenamiento.
- Es sencilla, compacta y no utiliza `Price` para construir la representación, por lo que no introduce fuga del objetivo por sí misma.
- Puede ser útil para modelos que aceptan variables numéricas y cuando la frecuencia de una categoría aporta señal.

**Desventajas**

- El conteo mide cuán frecuente es el suburbio, no cuánto cuestan sus propiedades. Un suburbio frecuente no necesariamente tiene precios altos.
- Dos suburbios diferentes pueden tener el mismo conteo y quedan representados por el mismo número, también se pierde la identidad de la categoría.
- Un modelo lineal puede interpretar diferencias entre conteos como una relación numérica ordenada que no necesariamente tiene sentido para la variable nominal.
- Los conteos dependen del tamaño y composición del entrenamiento. Un suburbio no observado recibe 0 en este ejercicio; esa regla debe definirse cuidadosamente para datos nuevos.
