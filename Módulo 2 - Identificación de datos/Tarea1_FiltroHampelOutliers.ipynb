## Investigación metodológica

El filtro de Hampel clásico es una regla robusta basada en la **mediana** y la **desviación absoluta mediana (MAD)**. Una forma univariada frecuente es:

$$M_i = \frac{0.6745(x_i - \operatorname{mediana}(x))}{MAD(x)}$$

Una observación se marca cuando $|M_i| > k$, donde $k$ debe justificarse. Para varias variables no existe una única extensión universal. En tu investigación debes elegir y explicar una de estas estrategias, o proponer otra con referencias:
