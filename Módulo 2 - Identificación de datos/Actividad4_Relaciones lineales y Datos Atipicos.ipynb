### Ejercicio 1. Relaciones lineales: Auto MPG

**Dataset:** [Auto MPG en UCI](https://archive.ics.uci.edu/dataset/9/auto+mpg) · [CSV alternativo reproducible](https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv)

1. Carga `mpg.csv` y documenta unidades, tipos, faltantes y duplicados.
2. Estudia la relación entre `horsepower` y `mpg`, y entre `weight` y `mpg` mediante diagramas de dispersión.
3. Calcula covarianza, correlación de Pearson y correlación de Spearman. Explica cuándo difieren.
4. Ajusta una regresión lineal simple para predecir `mpg` usando `weight`. Reporta pendiente, intercepto, $R^2$, MAE y RMSE.
5. Usa `origin` o `cylinders` para colorear la gráfica. Describe si parece haber grupos con rectas o niveles diferentes.
6. Detecta observaciones influyentes o extremas y repite el ajuste con una estrategia documentada. Compara los coeficientes y errores antes/después.
7. Explica por qué una correlación alta no demuestra que reducir el peso cause automáticamente mayor rendimiento.

### Ejercicio 2. Datos atípicos: California Housing

**Dataset:** [California Housing en scikit-learn](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.fetch_california_housing.html)

1. Descarga el dataset con `fetch_california_housing(as_frame=True)` y registra la fuente, tamaño y variables.
2. Usa IQR y Z-score para estudiar `MedInc`, `AveRooms` y `AveOccup`. Compara el número de candidatos.
3. Aplica una transformación o escalamiento robusto cuando sea necesario. Explica por qué los métodos sensibles a escala pueden cambiar.
4. Aplica **LOF** o **Isolation Forest** sobre al menos tres variables numéricas. Declara `contamination`, semilla y demás parámetros.
5. Compara candidatos univariados y multivariados. Identifica al menos tres registros detectados por un método y no por otro.
