## 1. ¿Qué es la Ingeniería de Características?

La **Ingeniería de Características** (*Feature Engineering*) es el proceso de usar conocimiento del dominio, estadística y transformaciones de datos para **crear, seleccionar y transformar variables (features)** que hagan que los modelos de Machine Learning funcionen mejor.

Una **característica** (*feature*) es una propiedad medible e individual de un fenómeno que se observa. Es la materia prima con la que aprenden los modelos.

>  **En términos simples:** la Ingeniería de Características es el arte y la ciencia de transformar datos crudos en variables que representen mejor el problema para los algoritmos de aprendizaje.

> *"Coming up with features is difficult, time-consuming, requires expert knowledge. Applied machine learning is basically feature engineering."* — Andrew Ng

### ¿Por qué es tan importante?

En la práctica, la calidad de las características suele tener **más impacto** en el desempeño de un modelo que la elección del algoritmo. Modelos simples con buenas características frecuentemente superan a modelos complejos con datos mal preparados.

$$\text{Datos crudos} \;\xrightarrow{\text{Ingeniería de Características}}\; \text{Representación útil} \;\rightarrow\; \text{Modelo}$$

### Historia breve de la Ingeniería de Características

| Época | Hito |
|---|---|
| **1960s–70s** | Estadística clásica: estandarización, transformaciones (log, Box-Cox) |
| **1980s** | Sistemas expertos — features diseñadas manualmente con reglas |
| **1990s** | ML estadístico: selección de variables, PCA para reducción |
| **2000s** | Auge de features manuales en visión y texto (SIFT, TF-IDF) |
| **2010s** | Kaggle populariza el *feature engineering* competitivo |
| **2012+** | Deep Learning aprende features automáticamente (*representation learning*) |
| **2015+** | AutoML y *featuretools* — generación automática de características |
| **2020+** | *Feature Stores* — gestión de features a escala en producción (MLOps) |



## 2. Importancia y Aplicaciones

La Ingeniería de Características es una etapa **transversal** en casi cualquier proyecto de ciencia de datos y determina, en gran medida, el éxito del modelo.

### ⭐ ¿Por qué es importante?

- **Mejora el desempeño**: buenas features aumentan accuracy, F1, AUC, etc.
- **Simplifica los modelos**: representaciones adecuadas permiten usar modelos más simples e interpretables.
- **Reduce el sobreajuste**: eliminar ruido y redundancia mejora la generalización.
- **Incorpora conocimiento del dominio**: traduce la experiencia experta en variables medibles.
- **Acelera el entrenamiento**: menos dimensiones y datos más limpios = cómputo más eficiente.
- **Mejora la interpretabilidad**: features con significado facilitan explicar decisiones del modelo.

### 🏦 Finanzas y Banca
- Scoring crediticio (ratios financieros, historial de pagos)
- Detección de fraude (frecuencia y monto de transacciones, features temporales)
- Predicción de *churn* (abandono de clientes)

- Detección de anomalías en líneas de producción
- Features de series de tiempo (medias móviles, rezagos)

### 🌐 Web y Tecnología
- Predicción de clics (CTR) en publicidad digital

- Motores de búsqueda y ranking

## 3. Conceptos Clave

### 📦 Característica (Feature)

Una **característica** (*feature*) es una variable de entrada que describe una propiedad medible del fenómeno de estudio. En una tabla, corresponde a una **columna**.

**Ejemplo:**
```
Registro: cliente con edad=34, ingreso=25000, ciudad="GDL"
Features:  edad (34), ingreso (25000), ciudad ("GDL")
Target:    ¿comprará? (Sí/No)
```

Las features son la **entrada** del modelo; la variable objetivo (*target*) es lo que se quiere predecir.

### 🏷️ Tipos de Variables

Identificar el tipo de cada variable es el **primer paso** para decidir qué transformación aplicar.

| Tipo | Descripción | Ejemplo |
|---|---|---|
| **Numérica continua** | Valores reales medibles | Ingreso, temperatura |
| **Numérica discreta** | Conteos enteros | Nº de hijos, visitas |
| **Categórica nominal** | Categorías sin orden | Ciudad, color |
| **Categórica ordinal** | Categorías con orden | Nivel educativo, talla (S/M/L) |
| **Binaria** | Dos valores posibles | ¿Fumador? (Sí/No) |
| **Temporal** | Fechas y tiempos | Fecha de compra |
| **Texto** | Cadenas no estructuradas | Comentarios, reseñas |

### 📚 Dataset

Un **dataset** es una colección estructurada de observaciones (**filas**) descritas por características (**columnas**). Se divide típicamente en:

| Conjunto | Uso |
|---|---|
| **Train** | Entrenar el modelo y ajustar las transformaciones |
| **Validation** | Ajustar hiperparámetros y evitar sobreajuste |
| **Test** | Evaluar el desempeño final de forma imparcial |

> ⚠️ **Regla de oro:** los parámetros de las transformaciones (medias, escalas, categorías) se aprenden **solo con el train** y luego se aplican al validation/test para evitar *data leakage* (fuga de información).

### 🧩 Elementos y problemas frecuentes en los datos

| Concepto | Definición |
|---|---|
| **Valores faltantes** | Datos ausentes (NaN, nulos) que deben imputarse o eliminarse |
| **Outliers** | Valores atípicos que distorsionan el modelo |
| **Cardinalidad** | Nº de categorías distintas de una variable |
| **Sesgo (skewness)** | Asimetría en la distribución de una variable numérica |
| **Multicolinealidad** | Alta correlación entre features que aporta redundancia |
| **Escala** | Rango y unidades muy distintos entre variables |
| **Desbalance** | Clases del target con proporciones muy dispares |

### La "maldición de la dimensionalidad"

A medida que aumenta el número de características, el espacio se vuelve más **disperso** y los modelos necesitan **exponencialmente** más datos para generalizar bien:

$$\text{Densidad de datos} \propto \frac{N}{k^{\,d}}$$

donde $N$ es el número de observaciones y $d$ el número de dimensiones (features).

> Por ello, **más features no siempre es mejor**: la selección y reducción de dimensionalidad son parte esencial de la Ingeniería de Características.



## 4. Elementos y Técnicas Clave

La Ingeniería de Características agrupa un conjunto de técnicas que se aplican según el **tipo de variable** y el **problema**. Estos son los elementos más importantes del curso.

### 🧹 Tratamiento de Datos Faltantes

Los valores ausentes deben manejarse antes de modelar:

| Estrategia | Descripción |
|---|---|
| **Eliminación** | Borrar filas o columnas con demasiados nulos |
| **Imputación simple** | Rellenar con media, mediana o moda |
| **Imputación avanzada** | KNN, MICE, modelos predictivos |
| **Indicador de faltante** | Crear una feature binaria "estaba ausente" |

### 🔤 Codificación de Variables Categóricas

Los modelos requieren números, no texto. Las categorías se transforman en variables numéricas:

| Técnica | Uso |
|---|---|
| **One-Hot Encoding** | Nominal, baja cardinalidad |
| **Label / Ordinal Encoding** | Categorías con orden |
| **Target / Mean Encoding** | Alta cardinalidad |
| **Frequency Encoding** | Reemplaza por frecuencia de aparición |

### 🔢 Transformación de Variables Numéricas

Ajustar escala y distribución mejora muchos algoritmos:

| Técnica | Objetivo |
|---|---|
| **Estandarización (Z-score)** | Media 0, desviación 1 |
| **Normalización (Min-Max)** | Escalar al rango [0, 1] |
| **Log / Box-Cox / Yeo-Johnson** | Reducir sesgo y estabilizar varianza |
| **Escalado robusto** | Resistente a outliers (usa mediana e IQR) |

Ejemplo de estandarización:

$$z = \frac{x - \mu}{\sigma}$$

### 🪜 Discretización (Binning)

Convertir variables continuas en intervalos o categorías:

- **Ancho fijo** (*equal-width*): intervalos del mismo tamaño.
- **Frecuencia igual** (*equal-frequency*): mismo número de datos por bin.
- **Basada en modelos** (árboles, K-means): cortes óptimos según los datos.

### ⏱️ Características en Series de Tiempo

Extraer información temporal para modelos predictivos:

- **Rezagos (*lags*)**: valor de $t-1$, $t-7$, etc.
- **Ventanas móviles**: medias, mínimos y máximos móviles.
- **Componentes de fecha**: día de la semana, mes, festivo.
- **Estacionalidad y tendencia**.

### 🎯 Selección de Características

Elegir el subconjunto de features más informativo:

| Enfoque | Ejemplo |
|---|---|
| **Filter** | Correlación, chi-cuadrado, información mutua |
| **Wrapper** | RFE (eliminación recursiva) |
| **Embedded** | Regularización L1 (Lasso), importancia de árboles |

### 📉 Reducción de Dimensionalidad

Comprimir muchas features en pocas componentes que preserven la información:

- **PCA** — Análisis de Componentes Principales.
- **t-SNE / UMAP** — visualización no lineal.

### ¿Para qué sirve todo esto?

- **Mejorar el desempeño** de cualquier modelo de ML.
- **Reducir ruido y redundancia** en los datos.
- **Facilitar la interpretación** de los resultados.
- **Optimizar el uso de recursos** de cómputo.

## 5. Pipeline Típico de Ingeniería de Características

Un proyecto de ciencia de datos sigue una secuencia de etapas donde la Ingeniería de Características es el **núcleo** de la preparación de datos. Este es el **pipeline estándar**:

```mermaid
flowchart TB
    P1["1 · Datos Crudos<br/><small>CSV, SQL, APIs, logs, sensores, texto</small>"]:::data
    P2["2 · Extracción y Adquisición<br/><small>Lectura de múltiples fuentes, integración</small>"]:::step
    P3["3 · Identificación / EDA<br/><small>Tipos de variables, distribuciones, outliers, correlaciones</small>"]:::step
    P4["4 · Limpieza<br/><small>Datos faltantes, outliers, inconsistencias</small>"]:::step
    P5["5 · Transformación / Creación de Features<br/><small>Codificación, escalado, discretización, series de tiempo, features derivadas</small>"]:::core
    P6["6 · Selección / Reducción<br/><small>Selección de características, reducción de dimensionalidad (PCA)</small>"]:::step
    P7["7 · Modelado<br/><small>Regresión, SVM, árboles, ensambles (RF, XGBoost)</small>"]:::model
    P8["8 · Evaluación y Despliegue<br/><small>Métricas, pipelines reproducibles, Feature Store / MLOps</small>"]:::model

    P1 --> P2 --> P3 --> P4 --> P5 --> P6 --> P7 --> P8

    classDef data fill:#f4d35e,stroke:#b8860b,stroke-width:1px,color:#1a3a5c;
    classDef step fill:#eaf2fb,stroke:#4a9eda,stroke-width:1px,color:#1a3a5c;
    classDef core fill:#2d6a9f,stroke:#1a3a5c,stroke-width:2px,color:#ffffff;
    classDef model fill:#d8f3dc,stroke:#2d6a4f,stroke-width:1px,color:#1a3a5c;
```

### Consideraciones importantes

1. **El orden importa**:  limpiar antes de transformar, ajustar solo con el *train*.
2. **Evitar *data leakage***:  nunca usar información del test al crear features.
3. **La calidad de los datos supera al modelo**:  "Garbage in, garbage out".
4. **El pipeline debe ser reproducible**:  usar `Pipeline` y `ColumnTransformer` de scikit-learn.
5. **Documentar cada transformación**:  facilita mantenimiento y auditoría.
