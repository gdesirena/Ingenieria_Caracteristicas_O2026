## 📖 Descripción de la Asignatura

La **ingeniería de características** es el proceso de seleccionar, transformar y crear variables (características) a partir de datos crudos para mejorar el rendimiento de los modelos de machine learning. Este proceso incluye tareas como la selección de las variables más relevantes, la creación de nuevas características a partir de las existentes, la conversión de datos categóricos a numéricos, el manejo de valores faltantes y la normalización o escalado de datos.

> Este curso proporciona los fundamentos teóricos y prácticos de la ingeniería de características (*feature engineering*) y está diseñado para que el estudiante desarrolle habilidades para identificar, transformar, crear y seleccionar características relevantes a partir de datos crudos, con el objetivo de mejorar el desempeño de los modelos predictivos.
