El conjunto de datos [Data/diamonds_faltantes.csv](Data/diamonds_faltantes.csv) contiene información de **más de 53,000 diamantes**, incluyendo atributos físicos, calidad gemológica y precio en el mercado. En este archivo se han introducido patrones de valores faltantes tanto en variables numéricas como categóricas para simular un escenario real de calidad de datos deficiente.

##### Diccionario de Variables:
- **`carat`**: Peso del diamante en quilates (cuantitativa continua).
- **`cut`**: Calidad del corte (`Fair`, `Good`, `Very Good`, `Premium`, `Ideal`) (cualitativa ordinal).
- **`color`**: Color del diamante, desde `J` (peor / amarillento) hasta `D` (mejor / incoloro) (cualitativa ordinal).
- **`clarity`**: Pureza del diamante (`I1`, `SI2`, `SI1`, `VS2`, `VS1`, `VVS2`, `VVS1`, `IF`) (cualitativa ordinal).
- **`depth`**: Porcentaje de profundidad total respecto al diámetro promedio (cuantitativa continua).
- **`table`**: Ancho de la tabla superior como porcentaje del ancho total (cuantitativa continua).
- **`price`**: Precio en dólares estadounidenses (USD) (variable objetivo cuantitativa).
- **`x`**: Longitud en milímetros (cuantitativa continua).
- **`y`**: Ancho en milímetros (cuantitativa continua).
- **`z`**: Profundidad en milímetros (cuantitativa continua).
