# Guía del proyecto final

El proyecto evalúa la aplicación de los conocimientos del curso a una base de datos asignada o elegida por el equipo. La meta es comprender los datos mediante un análisis exploratorio (EDA), preparar variables, explorar agrupamientos y resolver una tarea de aprendizaje automático.


## Objetivo general

Realizar un EDA y construir un proceso reproducible que incluya técnicas de ingeniería de características, análisis de clustering y un modelo supervisado adecuado al problema. Presenten los resultados con gráficas, comentarios y conclusiones basadas en evidencia.

## 1. Comprender los datos y definir el problema

Antes de transformar variables, describan el conjunto de datos y el fenómeno que representa. Delimiten una pregunta que pueda responderse con la información disponible.

- Identifiquen la fuente, fecha o periodo, unidad de observación y responsable de los datos.
- Expliquen qué significa cada variable relevante, sus unidades y su tipo.
- Si el equipo elige la base, justifiquen su pertinencia y mencionen sus limitaciones.
- Definan qué resultado esperan obtener del análisis y quién podría utilizarlo.

### Análisis exploratorio (EDA)

Revisen dimensiones, tipos de datos, duplicados, valores faltantes, rangos y categorías. Analicen distribuciones, posibles valores atípicos y relaciones entre variables mediante tablas y visualizaciones. Cada figura debe ayudar a responder una pregunta; acompañen los hallazgos con una interpretación, no solo con una descripción visual.

## 2. Preparar e ingeniería de características

Apliquen las técnicas vistas en los tres parciales según las necesidades del problema. Para cada decisión, indiquen qué variable se modifica, por qué eligieron el método y cómo cambia su interpretación.

1. **Identificación de datos:** clasifiquen las variables por tipo, escala y función.
2. **Datos faltantes:** analicen su patrón y justifiquen si imputan, descartan o conservan valores ausentes.
3. **Variables categóricas:** seleccionen una codificación coherente con la naturaleza de cada categoría.
4. **Variables numéricas:** consideren transformaciones para asimetrías, distribuciones o relaciones no lineales.
5. **Escalamiento:** expliquen si el algoritmo requiere variables comparables en escala.
6. **Discretización:** justifiquen los intervalos y revisen la información que se pierde al agrupar valores.

No apliquen transformaciones de forma automática. Para evitar fuga de información, ajusten imputadores, escaladores y otros transformadores solo con datos de entrenamiento; luego apliquen lo aprendido al conjunto de prueba.

## 3. Explorar agrupamientos

A partir de los datos preparados, realicen clustering y describan qué significa cada grupo en el contexto del problema. El PDF solicita los siguientes métodos:

- **K-means++:** utilicen K-means con inicialización `k-means++`. Comparen varios valores de $k$, justifiquen la selección y perfilen los grupos.
- **Clustering jerárquico:** documenten la distancia y el criterio de enlace. Cuando sea útil, incluyan e interpreten un dendrograma.

Muestren gráficas legibles y comenten tamaños, diferencias y limitaciones de los segmentos. Expliquen cómo evaluaron si los grupos son interpretables y útiles; no concluyan que son relevantes únicamente porque el algoritmo los produjo.

> **Importante:** un árbol de decisión no es un algoritmo de clustering. Es un método supervisado. Si lo utilizan para describir qué variables diferencian los grupos, preséntenlo como perfilado posterior de segmentos, no como un tercer algoritmo de agrupamiento.

## 4. Entrenar un modelo supervisado

Definan la **variable de salida** que desean predecir y las **variables de entrada** disponibles al momento de hacer la predicción. Justifiquen esta elección según el problema y excluyan columnas que revelen directamente la respuesta o información que no estaría disponible en el uso real.

Seleccionen un algoritmo de aprendizaje supervisado apropiado, por ejemplo, un árbol de decisión o un modelo de regresión. Establezcan una línea base y separen los datos en entrenamiento y prueba antes de ajustar transformaciones o modelos. Usen partición estratificada cuando corresponda.

El PDF solicita obtener un *accuracy* superior al 80 %. Esta meta corresponde a clasificación. Si las clases están desbalanceadas, reporten también precisión, *recall*, F1 y matriz de confusión. Si el problema es de regresión, *accuracy* no es adecuado: elijan y expliquen métricas como MAE, RMSE o $R^2$.

## 5. Evaluar, interpretar y mejorar

Evalúen el modelo en datos que no se usaron para entrenarlo. Reporten las métricas junto con el contexto, los errores observados y las limitaciones. No cambien la partición de prueba ni permitan que información de prueba influya en las transformaciones.

Si no se alcanza la referencia de desempeño, el PDF indica volver al primer punto del proceso. Interpreten esto como un ciclo de mejora: revisen la calidad de los datos, el objetivo, las características, el desbalance y las decisiones del modelo; documenten qué cambiaron y por qué. No modifiquen el análisis solo para elevar una métrica sin justificarlo.

Cierren cada etapa con evidencia: qué hicieron, qué resultado obtuvieron, cómo lo interpretan y qué decisión sigue.

## 6. Entregables y contenido del reporte

El proyecto requiere tres entregables:

1. **Presentación con diapositivas:** problema, enfoque, resultados relevantes y conclusiones.
2. **Reporte digital:** documento elaborado en computadora, no fotografías de hojas o cuadernos.
3. **Notebook de Jupyter:** código del proceso, analíticas, comentarios y gráficas, organizado para que otra persona pueda seguir el análisis.

El reporte debe incluir como mínimo:

- Nombres y apellidos de todas las personas integrantes del equipo.
- El código generado para realizar el proyecto, incluido en el reporte o entregado como un anexo claramente identificado.
- Un capítulo que describa la solución propuesta para el problema.
- Figuras que permitan visualizar los resultados.
- Una sección de conclusiones.

No es necesario imprimir a color. Diseñen las figuras para que sigan siendo distinguibles en escala de grises, usando, por ejemplo, líneas, marcadores o patrones diferentes.

## 7. Conclusiones y revisión final

Las conclusiones deben responder con evidencia a preguntas como estas:

- ¿Qué se aprendió sobre los datos y el fenómeno estudiado?
- ¿Qué transformaciones fueron relevantes y cómo afectaron el análisis?
- ¿Qué características distinguen los grupos y qué utilidad tienen?
- ¿Qué desempeño obtuvo el modelo en datos de prueba y qué errores cometió?
- ¿Cuáles son las limitaciones y qué trabajo futuro recomiendan?
